```python
class Rekening:
  def __init__(self, saldo_awal):
    self.__saldo = saldo_awal
  def __catat(self, pesan):
    print(f"LOG: {pesan}")
  def setor(self, jumlah):
    if jumlah <= 0:
      raise ValueError("Jumlah setoran harus positif")
    self.__saldo += jumlah
    self.__catat(f"setor {jumlah}")
  def lihat_saldo(self):
    return self.__saldo
```

```python
rekening1 = Rekening(100000)
print("Saldo awal:", rekening1.lihat_saldo())
```

```python
rekening1.setor(50000)
print("Saldo setelah setor:", rekening1.lihat_saldo())

rekening1.setor(40000)
rekening1.setor(30000)
print("Saldo setelah setor:", rekening1.lihat_saldo())
```
