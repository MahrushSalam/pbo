```python
class motor:
  def __init__(self, rpm):
    self.rpm = rpm

  @property
  def rpm(self):
    return self.__rpm

  @rpm.setter
  def rpm(self, nilai):
    if nilai < 0:
        raise ValueError("RPM tidak boleh negatif")
    else:
      self.__rpm = nilai

motor = motor(1500)
motor.rpm = 1800
print(motor.rpm)
```

```python
class suhu:
  def __init__(self, celcius):
    self.celcius = celcius

  @classmethod
  def dari_fahrenheit(cls, fahrenheit):
    celcius = (fahrenheit - 32) * 5/9
    return cls(celcius)

suhu = suhu.dari_fahrenheit(100)
print(suhu.celcius)
```
